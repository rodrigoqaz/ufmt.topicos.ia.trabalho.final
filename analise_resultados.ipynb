```python
import pandas as pd
import glob

# 1. Lista com os nomes dos arquivos (conforme sua imagem)
arquivos = [
    "resultados_baseline.csv",
    "resultados_full_lora.csv",
    "resultados_full_training.csv",
    "resultados_last_layer.csv"
]

# 2. Carregar e concatenar todos os arquivos em um único DataFrame
dfs = []
for arquivo in arquivos:
    try:
        df_temp = pd.read_csv(arquivo)
        dfs.append(df_temp)
    except FileNotFoundError:
        print(f"Aviso: O arquivo {arquivo} não foi encontrado.")

if dfs:
    df_geral = pd.concat(dfs, ignore_index=True)

    # 3. Realizar o Pivot para transformar os experimentos em colunas
    # Usamos 'Abstract' e 'Titulo_Real' como identificadores únicos (índice)
    # A coluna 'Experimento' vai virar o cabeçalho das novas colunas
    # Os valores serão preenchidos com o 'Titulo_Predito'
    df_pivot = df_geral.pivot_table(
        index=['Abstract', 'Titulo_Real'], 
        columns='Experimento', 
        values='Titulo_Predito',
        aggfunc='first' # Pega o primeiro valor encontrado (caso haja duplicatas)
    ).reset_index()

    # 4. Renomear as colunas para o formato desejado (Titulo_Predito_NOME)
    # O nome das colunas de experimentos estará no nível superior, vamos ajustá-los
    novos_nomes = {col: f"Titulo_Predito_{col}" for col in df_pivot.columns if col not in ['Abstract', 'Titulo_Real']}
    df_final = df_pivot.rename(columns=novos_nomes)

    # Remove o nome da variável de colunas 'Experimento' para deixar a tabela limpa
    df_final.columns.name = None

    # 5. Exibir e Salvar o resultado
    print("Visualização das primeiras linhas:")
    display(df_final.head(8))
    
    # Salvar em um novo CSV comparativo
    df_final.head(8).to_csv("comparacao_experimentos.csv", index=False)
    print("\nArquivo 'comparacao_experimentos.csv' salvo com sucesso!")

else:
    print("Nenhum arquivo foi carregado.")
```
